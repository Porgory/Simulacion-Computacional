# Confidence intervals

```python
import numpy as np
from scipy import stats
from IPython.display import display, Markdown
```

```python
# Medidas de desempeño pero ahora para 200
data = {
    "Tiempo promedio en el sistema": [0.0250, 0.0304, 0.0202, 0.0207, 0.0284, 0.0271, 0.0301, 0.0304, 0.0211, 0.0236],
    "Porcentaje de tiempo idle": [0.52, 0.51, 0.59, 0.51, 0.57, 0.45, 0.50, 0.47, 0.55, 0.50],
    "Tiempo promedio de espera por cliente": [0.0200, 0.0326, 0.0321, 0.0252, 0.0172, 0.0307, 0.0391, 0.0149, 0.0350, 0.0170],
    "Fraccion que tuvo que esperar": [0.25, 0.34, 0.32, 0.23, 0.25, 0.34, 0.37, 0.26, 0.32, 0.26],
    "Tiempo promedio de espera de los que esperaron": [0.0800, 0.0951, 0.1000, 0.1079, 0.0687, 0.0914, 0.1060, 0.0585, 0.1102, 0.0666],
}
```

```python
def calcular_ic(M, alpha, nombre_medida):
    reps = len(M)
    confidence = (1 - alpha) * 100

    mean = np.mean(M)
    desv = np.std(M, ddof=1)  # ddof=1 => desviación estándar muestral

    tval = stats.t.ppf(1 - (alpha/2), reps - 1)
    hval = tval * (desv / np.sqrt(reps))

    display(Markdown(
        rf"""
**{nombre_medida} — {confidence:.0f}% de confianza**

$\bar{{X}} = {mean:.4f}$

$S = {desv:.4f}$

$t_{{n-1,1-\alpha/2}} = t_{{{reps-1}, {1-alpha/2}}} = {tval:.4f}$

$h = {hval:.4f}$

$C.I. = (\bar{{X}}-h,\ \bar{{X}}+h) = ({mean-hval:.4f},\ {mean+hval:.4f})$

---
"""))
    return mean, desv, hval
```

```python
resultados = {}

for nombre, valores in data.items():
    resultados[nombre] = {}
    for alpha in [0.05, 0.01]:
        mean, desv, hval = calcular_ic(valores, alpha, nombre)
        resultados[nombre][1-alpha] = (mean, desv, hval)
```
