5.1.1. MidSquare
Tomando en cuenta el tema visto en clase en las diapositivas de Números Aleatorios implemente o use un
generador con el método de MidSquare o Middle-Square (o cuadrados medios) para generar una secuencia ui
= u1, u2, ..., un de n =100 números pseudoaleatorios usando las siguientes semillas:
a. 1931
b. 675248
c. 4567
d. 12122
e. 7179345

```python
def mid_square(seed, cantidad=100):
    # Cantidad de cifras de la semilla
    digitos = len(str(seed))

    # La semilla debe conservar exactamente sus cifras
    x = str(seed).zfill(digitos)

    resultados = []

    for i in range(cantidad):

        # -----------------------------------------
        # 1. Elevar x al cuadrado
        # -----------------------------------------
        cuadrado = int(x) ** 2

        # -----------------------------------------
        # 2. Completar con ceros a la izquierda
        #    hasta tener 4n cifras
        # -----------------------------------------
        cuadrado = str(cuadrado).zfill(4 * digitos)

        # -----------------------------------------
        # 3. Extraer las 2n cifras centrales
        # -----------------------------------------
        inicio = digitos
        fin = 3 * digitos

        x = cuadrado[inicio:fin]

        # -----------------------------------------
        # 4. Formar ui
        #    Ejemplo: x = 3728 -> u = 0.3728
        # -----------------------------------------
        u = "0." + x

        resultados.append((i + 1, x, u))

    return resultados


# =============================================
# SEMILLAS DEL EJERCICIO
# =============================================

semillas = [1931, 675248, 4567, 12122, 7179345]


# =============================================
# MOSTRAR RESULTADOS
# =============================================

for semilla in semillas:

    print("\n" + "=" * 50)
    print("SEMILLA:", semilla)
    print("=" * 50)

    resultados = mid_square(semilla, 100)

    for i, x, u in resultados:
        print(f"{i:3d} | x = {x} | u = {u}")
```

5.1.2. Generador congruencial mixto
Tomando en cuenta el tema visto en clase en las diapositivas de Números Aleatorios y el Capítulo 3 Números
Aleatorios de [Ross99], implemente o use el método de generador congruencial mixto o generador lineal
congruencial (GLC) de acuerdo con la siguiente ecuación:

Para generar una secuencia ui = u1, u2, ..., un de n=100 números pseudoaleatorios usando los siguientes
parámetros a, c, m del GLC con su respectiva semilla X0 :
a. X0 = 127, a = 115, c = 0, m = 128
b. X0 = 43, a = 25, c = 5, m = 48
c. X0 = 13, a = 16807, c = 0, m = 231 – 1
d. X0 = 29022024, a = 25214903917, c = 11, m = 248
e. X0 = 3, a = 5, c = 7, m = 200

```python
def genran(a, c, m, xant):
    # Fórmula del Generador Lineal Congruencial
    xsig = (a * xant + c) % m

    # Número pseudoaleatorio entre 0 y 1
    usig = xsig / m

    return [xsig, usig]


def genranN(a, c, m, x0, N):
    xant = x0

    I = []
    X = []
    U = []

    for t in range(1, N + 1):

        [xi, ui] = genran(a, c, m, xant)

        I.append(t)
        X.append(xi)
        U.append(ui)

        xant = xi

    return [I, X, U]


def showValues(I, X, U):
    print("i\tXi\tUi")

    for t in range(len(X)):
        print("%d\t%d\t%.6f" % (I[t], X[t], U[t]))


# ============================================
# CANTIDAD DE NÚMEROS
# ============================================

N = 100


# ============================================
# INCISO A
# X0 = 127, a = 115, c = 0, m = 128
# ============================================

print("\n==========================================")
print("INCISO A")
print("==========================================")

x0 = 127
a = 115
c = 0
m = 128

[I, X, U] = genranN(a, c, m, x0, N)

showValues(I, X, U)


# ============================================
# INCISO B
# X0 = 43, a = 25, c = 5, m = 48
# ============================================

print("\n==========================================")
print("INCISO B")
print("==========================================")

x0 = 43
a = 25
c = 5
m = 48

[I, X, U] = genranN(a, c, m, x0, N)

showValues(I, X, U)


# ============================================
# INCISO C
# X0 = 13, a = 16807, c = 0, m = 2^31 - 1
# ============================================

print("\n==========================================")
print("INCISO C")
print("==========================================")

x0 = 13
a = 16807
c = 0
m = 2**31 - 1

[I, X, U] = genranN(a, c, m, x0, N)

showValues(I, X, U)


# ============================================
# INCISO D
# X0 = 29022024
# a = 25214903917
# c = 11
# m = 2^48
# ============================================

print("\n==========================================")
print("INCISO D")
print("==========================================")

x0 = 29022024
a = 25214903917
c = 11
m = 2**48

[I, X, U] = genranN(a, c, m, x0, N)

showValues(I, X, U)


# ============================================
# INCISO E
# X0 = 3, a = 5, c = 7, m = 200
# ============================================

print("\n==========================================")
print("INCISO E")
print("==========================================")

x0 = 3
a = 5
c = 7
m = 200

[I, X, U] = genranN(a, c, m, x0, N)

showValues(I, X, U)
```

5.1.3. Otros generadores de números aleatorios
Tomando en cuenta la lectura del artículo científico [Mancilla2000], implemente al menos dos (2) métodos de los
siguientes generadores de numeros pseudoaleatorios y genere una secuencia ui = u1, u2, ..., un de n =100
números aleatorios con unos parámetros de ejemplo que Ud. escoja.
a. Generador Shift-Register
b. Generador Lagged-Fibonnaci
c. Generador de Congruencia Inversa
d. Generador de Congruencia Lineal Combinada

```python
def shift_register(n):
    x = 123456789
    numeros = []

    for i in range(n):
        x ^= (x << 13)
        x ^= (x >> 17)
        x ^= (x << 5)
        x &= 0xFFFFFFFF

        u = x / 2**32
        numeros.append(u)

    return numeros


def lagged_fibonacci(n):
    j = 24
    k = 55
    semilla = 123456789
    m = 2**31

    x = []

    for i in range(k):
        semilla = (1103515245 * semilla + 12345) % m
        x.append(semilla)

    for i in range(n):
        nuevo = (x[i + k - j] + x[i]) % m
        x.append(nuevo)

    numeros = []

    for i in range(n):
        valor = x[k + i] / m
        numeros.append(valor)

    return numeros


def congruencia_inversa(n):
    x = 7
    a = 3
    b = 5
    m = 101

    numeros = []

    for i in range(n):
        if x == 0:
            inverso = 0
        else:
            inverso = pow(x, -1, m)

        x = (a * inverso + b) % m
        u = x / m

        numeros.append(u)

    return numeros


def congruencia_lineal_combinada(n):
    x = 7
    a1 = 23
    c1 = 17
    m1 = 101

    y = 11
    a2 = 31
    c2 = 13
    m2 = 103

    numeros = []

    for i in range(n):
        x = (a1 * x + c1) % m1
        y = (a2 * y + c2) % m2

        z = (x + y) % max(m1, m2)
        u = z / max(m1, m2)

        numeros.append(u)

    return numeros


n = 100

numeros_shift_register = shift_register(n)
numeros_lagged_fibonacci = lagged_fibonacci(n)
numeros_congruencia_inversa = congruencia_inversa(n)
numeros_congruencia_combinada = congruencia_lineal_combinada(n)


print("A. GENERADOR SHIFT-REGISTER")

for i in range(n):
    print("u" + str(i + 1) + " =", round(numeros_shift_register[i], 6))


print("\nB. GENERADOR LAGGED-FIBONACCI")

for i in range(n):
    print("u" + str(i + 1) + " =", round(numeros_lagged_fibonacci[i], 6))


print("\nC. GENERADOR DE CONGRUENCIA INVERSA")

for i in range(n):
    print("u" + str(i + 1) + " =", round(numeros_congruencia_inversa[i], 6))


print("\nD. GENERADOR DE CONGRUENCIA LINEAL COMBINADA")

for i in range(n):
    print("u" + str(i + 1) + " =", round(numeros_congruencia_combinada[i], 6))
```

5.1.4. Tamaño de ciclo, pruebas de uniformidad y aleatoriedad
Tomando en cuenta el tema visto en clase en las diapositivas de Números Aleatorios y Test o contrastes para
números aleatorios, implementando o usando los respectivos códigos de Jupyter Notebook en Python (.ipynb),
calcule para cada una de las secuencias de números pseudoaleatorios generados en los puntos 5.1.1., 5.1.2., y
5.1.3. lo siguiente:a. Tamaño de ciclo o periodo de la secuencia si lo tiene.
b. Prueba de uniformidad con alguno de los métodos vistos en clase
c. Prueba de aleatoriedad con el método visto en clase
De acuerdo con el tamaño de ciclo encontrado y las pruebas de uniformidad y aleatoriedad, responda:
• ¿Cuáles son los peores generadores de números aleatorios y porqué?
• ¿Cuáles son los mejores generadores de números aleatorios y porqué?

```python
import numpy as np
import pandas as pd
from scipy.stats import chisquare, norm

# ---------------------------------------------------------
# 1) Tamaño de ciclo / periodo de una secuencia
# ---------------------------------------------------------
def periodo_secuencia(u):
    vistos = {}
    for i, valor in enumerate(u):
        clave = round(float(valor), 15)
        if clave in vistos:
            return i - vistos[clave]
        vistos[clave] = i
    return len(u)


# ---------------------------------------------------------
# 2) Prueba de uniformidad: Chi-cuadrado
# ---------------------------------------------------------
def prueba_uniformidad(u, bins=10):
    datos = np.asarray(u, dtype=float)
    counts, _ = np.histogram(datos, bins=bins, range=(0, 1))
    esperado = len(datos) / bins
    chi2, p_valor = chisquare(counts, f_exp=np.full(bins, esperado))
    return {
        "bins": bins,
        "conteos": counts.tolist(),
        "esperado": esperado,
        "chi2": chi2,
        "p_value": p_valor
    }


# ---------------------------------------------------------
# 3) Prueba de aleatoriedad: Runs test (sobre la mediana)
# ---------------------------------------------------------
def prueba_aleatoriedad(u):
    datos = np.asarray(u, dtype=float)
    mediana = np.median(datos)
    serie = np.where(datos >= mediana, 1, 0)

    runs = 1
    for i in range(1, len(serie)):
        if serie[i] != serie[i - 1]:
            runs += 1

    n1 = np.sum(serie == 1)
    n2 = np.sum(serie == 0)

    if n1 == 0 or n2 == 0:
        return {
            "runs": runs,
            "z": 0.0,
            "p_value": 1.0,
            "n1": n1,
            "n2": n2
        }

    media = 2 * n1 * n2 / (n1 + n2) + 1
    varianza = 2 * n1 * n2 * (2 * n1 * n2 - n1 - n2) / (((n1 + n2) ** 2) * (n1 + n2 - 1))
    z = (runs - media) / np.sqrt(varianza)
    p_valor = 2 * norm.sf(abs(z))

    return {
        "runs": runs,
        "z": z,
        "p_value": p_valor,
        "n1": n1,
        "n2": n2
    }


# ---------------------------------------------------------
# 4) Secuencias de los puntos anteriores
# ---------------------------------------------------------
# 5.1.1 MidSquare

def secuencia_mid_square(seed, n=100):
    x = abs(int(seed))
    if x == 0:
        return [0.0]
    d = len(str(x))
    valores = []
    vistos = set()
    while len(valores) < n:
        cuadrado = str(x * x).zfill(2 * d)
        inicio = (len(cuadrado) - d) // 2
        x = int(cuadrado[inicio:inicio + d])
        if x == 0 or x in vistos:
            break
        vistos.add(x)
        valores.append(round(x / (10 ** d), 10))
    return valores


# 5.1.2 Generador congruencial lineal
# Las funciones ya fueron definidas en la celda anterior

def secuencia_glc(x0, a, c, m, n=100):
    xant = x0
    u = []
    for _ in range(n):
        xant = (a * xant + c) % m
        u.append(xant / m)
    return u


# 5.1.3 Otros generadores
# Las funciones ya existen en la celda anterior

def secuencia_shift_register(n=100):
    x = 123456789
    nums = []
    for _ in range(n):
        x ^= (x << 13)
        x ^= (x >> 17)
        x ^= (x << 5)
        x &= 0xFFFFFFFF
        nums.append(x / 2**32)
    return nums


def secuencia_lagged_fibonacci(n=100):
    j = 24
    k = 55
    semilla = 123456789
    m = 2**31
    x = []
    for _ in range(k):
        semilla = (1103515245 * semilla + 12345) % m
        x.append(semilla)
    for i in range(n):
        nuevo = (x[i + k - j] + x[i]) % m
        x.append(nuevo)
    out = []
    for i in range(n):
        out.append(x[k + i] / m)
    return out


def secuencia_congruencia_inversa(n=100):
    x = 7
    a = 3
    b = 5
    m = 101
    nums = []
    for _ in range(n):
        inverso = 0 if x == 0 else pow(x, -1, m)
        x = (a * inverso + b) % m
        nums.append(x / m)
    return nums


def secuencia_congruencia_lineal_combinada(n=100):
    x = 7
    a1 = 23
    c1 = 17
    m1 = 101
    y = 11
    a2 = 31
    c2 = 13
    m2 = 103
    nums = []
    for _ in range(n):
        x = (a1 * x + c1) % m1
        y = (a2 * y + c2) % m2
        z = (x + y) % max(m1, m2)
        nums.append(z / max(m1, m2))
    return nums


# ---------------------------------------------------------
# 5) Construcción de la tabla resumen
# ---------------------------------------------------------
secuencias = {}
for semilla in [1931, 675248, 4567, 12122, 7179345]:
    secuencias[f"MidSquare_{semilla}"] = secuencia_mid_square(semilla, 100)

for nombre, params in {
    "GLC_A": (127, 115, 0, 128),
    "GLC_B": (43, 25, 5, 48),
    "GLC_C": (13, 16807, 0, 2**31 - 1),
    "GLC_D": (29022024, 25214903917, 11, 2**48),
    "GLC_E": (3, 5, 7, 200),
}.items():
    x0, a, c, m = params
    secuencias[nombre] = secuencia_glc(x0, a, c, m, 100)

secuencias["ShiftRegister"] = secuencia_shift_register(100)
secuencias["LaggedFibonacci"] = secuencia_lagged_fibonacci(100)
secuencias["CongruenciaInversa"] = secuencia_congruencia_inversa(100)
secuencias["CongruenciaLinealCombinada"] = secuencia_congruencia_lineal_combinada(100)

resumen = []
for nombre, u in secuencias.items():
    unif = prueba_uniformidad(u, bins=10)
    alea = prueba_aleatoriedad(u)
    resumen.append({
        "Generador": nombre,
        "Periodo": periodo_secuencia(u),
        "Chi2": round(unif["chi2"], 6),
        "p_uniformidad": round(unif["p_value"], 6),
        "Runs_z": round(alea["z"], 6),
        "p_aleatoriedad": round(alea["p_value"], 6),
    })

resumen_df = pd.DataFrame(resumen)
print(resumen_df.to_string(index=False))

# ---------------------------------------------------------
# 6) Interpretación
# ---------------------------------------------------------
print("\nConclusión rápida:")
print("- Los peores generadores suelen ser los de periodo corto, m pequeño o estructuras con repeticiones tempranas.")
print("- Los mejores generadores son los que tienen módulo grande y periodo largo, como GLC con m = 2^31 - 1 o m = 2^48.")
print("- El MidSquare es especialmente débil porque el cuadrado centrado puede producir ciclos muy cortos y comportamientos sesgados.")

# ---------------------------------------------------------
# 7) Ejemplo de respuesta del laboratorio
# ---------------------------------------------------------
respuesta = """
Para cada secuencia se calcula el periodo, la prueba de uniformidad (chi-cuadrado) y la prueba de aleatoriedad (runs test).
Los peores generadores son los que presentan ciclos cortos, módulos pequeños y repeticiones tempranas, como MidSquare y algunos GLC con m pequeño.
Los mejores son los generadores con modulo grande y periodo largo, como el GLC con m = 2^31 - 1 o m = 2^48, porque mantienen mejor la apariencia uniforme y más aleatoria.
"""
print("\n" + respuesta)
```
