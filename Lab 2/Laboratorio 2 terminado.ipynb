```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.integrate import quad
from math import exp, sqrt, pi

# Semilla para que los resultados sean reproducibles
SEED = 2026
rng = np.random.default_rng(SEED)

# Número de observaciones de Monte Carlo
N = 1_000_000

print(f"N = {N:,}")
print(f"Semilla = {SEED}")

def monte_carlo_finito(g, a, b, n=N, rng=rng):
    u = rng.random(n)
    x = a + (b - a) * u
    valores = g(x)
    estimacion = (b - a) * np.mean(valores)
    error_estandar = (b - a) * np.std(valores, ddof=1) / np.sqrt(n)
    return estimacion, error_estandar

def resultado(nombre, estimacion, exacto=None, error_estandar=None):
    fila = {
        "Ejercicio": nombre,
        "Estimación Monte Carlo": estimacion,
        "Valor exacto": exacto,
        "Error absoluto": abs(estimacion - exacto) if exacto is not None else np.nan,
        "Error estándar": error_estandar
    }
    return fila
```

```python
#Ejercicio 3
g3 = lambda x: np.exp(np.exp(x))

est3, se3 = monte_carlo_finito(g3, 0, 1)
exact3, _ = quad(lambda x: np.exp(np.exp(x)), 0, 1)

r3 = resultado("3", est3, exact3, se3)
pd.DataFrame([r3])

#Ejercicio 4
g4 = lambda x: (1 - x**2)**(3/2)

est4, se4 = monte_carlo_finito(g4, 0, 1)
exact4 = 3 * pi / 16

r4 = resultado("4", est4, exact4, se4)
pd.DataFrame([r4])

#Ejercicio 5

g5 = lambda x: np.exp(x + x**2)

est5, se5 = monte_carlo_finito(g5, -2, 2)
exact5, _ = quad(lambda x: np.exp(x + x**2), -2, 2)

r5 = resultado("5", est5, exact5, se5)
pd.DataFrame([r5])

#Ejercicio 6

def h6(y):
    x = (1 - y) / y
    return x * (1 + x**2)**(-2) / y**2

u6 = rng.random(N)
val6 = h6(u6)

est6 = np.mean(val6)
se6 = np.std(val6, ddof=1) / np.sqrt(N)
exact6 = 0.5

r6 = resultado("6", est6, exact6, se6)
pd.DataFrame([r6])

#Ejercicio 7

def h7(y):
    x = (1 - y) / y
    return np.exp(-x**2) / y**2

u7 = rng.random(N)
val7 = h7(u7)

est7 = 2 * np.mean(val7)
se7 = 2 * np.std(val7, ddof=1) / np.sqrt(N)
exact7 = sqrt(pi)

r7 = resultado("7", est7, exact7, se7)
pd.DataFrame([r7])

#Ejercicio 8

u8 = rng.random(N)
v8 = rng.random(N)

x8 = u8
y8 = 2 * v8

val8 = np.exp(x8 + y8)

est8 = 2 * np.mean(val8)
se8 = 2 * np.std(val8, ddof=1) / np.sqrt(N)
exact8 = (np.e - 1) * (np.e**2 - 1)

r8 = resultado("8", est8, exact8, se8)
pd.DataFrame([r8])

#Ejercicio 9

u9 = rng.random(N)
v9 = rng.random(N)

x9 = (1 - u9) / u9
y9 = (1 - v9) / v9

indicador9 = (y9 < x9).astype(float)

val9 = (
    np.exp(-(x9 + y9))
    * indicador9
    / (u9**2 * v9**2)
)

est9 = np.mean(val9)
se9 = np.std(val9, ddof=1) / np.sqrt(N)

# Valor exacto: integral_0^inf e^(-x) [1-e^(-x)] dx = 1/2
exact9 = 0.5

r9 = resultado("9", est9, exact9, se9)
pd.DataFrame([r9])

#Comparacion de resultados

resultados = pd.DataFrame([r3, r4, r5, r6, r7, r8, r9])

pd.set_option("display.float_format", lambda x: f"{x:.10f}")
resultados
```

#5.3 GENERACIÓN DE VARIABLES ALEATORIAS DISCRETAS
Tomando en cuenta el tema visto en clase en las diapositivas de Generación de Variables Aleatorias Discretas
y el Capítulo 4 Generación de variables aleatorias discretas de [Ross99], implemente o use el método de la
transformada inversa para generar los primeros 100 valores aleatorios de las siguientes distribuciones de probabilidad, usando el generador congruencial mixto con los siguientes parámetros: X0 = 2391, a =
25214903917, c = 11, m = 248
.
a. p1 = 0.20, p2 = 0.15, p3 = 0.25, p4 = 0.40, donde pj = P{X = j}
b. P{X = 1} = 0.3, P{X = 2} = 0.2, P{X = 3} = 0.35, P{X = 4} = 0.15
c. Escriba un programa para generar n valores a partir de la función de masa de probabilidad p1=1/3, p2=2/3.
- Sea n = 100, ejecute el programa y determine la proporción de valores que sean iguales a 1.
- Repita ahora con n=1000
- Repita ahora con n=10000.
d. Suponga que la variable aleatoria X puede tomar cualquiera de los valores 1, …, 10 con probabilidades
respectivas 0.06, 0.06, 0.06, 0.06, 0.06, 0.15, 0.13, 0.14, 0.15, 0.13. Utilice el método de composición para
dar un algoritmo que genere el valor de X.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import comb, sqrt

X0, a, c, m = 2391, 25214903917, 11, 2**48

def glc(n, x0=X0, a=a, c=c, m=m):
    x = x0
    xs, us = [], []
    for _ in range(n):
        x = (a*x + c) % m
        xs.append(x)
        us.append(x/m)
    return np.array(xs), np.array(us)

def transformada_inversa(u, valores, probabilidades):
    valores = np.asarray(valores)
    probabilidades = np.asarray(probabilidades, dtype=float)
    if not np.isclose(probabilidades.sum(), 1):
        raise ValueError("Las probabilidades deben sumar 1.")
    acumuladas = np.cumsum(probabilidades)
    return valores[np.searchsorted(acumuladas, u, side="left")]

def tabla_frecuencias(muestra, valores):
    f = pd.Series(muestra).value_counts().reindex(valores, fill_value=0)
    return pd.DataFrame({
        "Valor": valores,
        "Frecuencia": f.values,
        "Proporción": f.values/len(muestra)
    })

_, u = glc(10)
pd.DataFrame({"i": range(1,11), "X_i": glc(10)[0], "U_i": u})
```

#Punto A


```python
_, u = glc(100)
valores = np.array([1,2,3,4])
prob = np.array([.20,.15,.25,.40])
muestra_a = transformada_inversa(u, valores, prob)
tabla_a = tabla_frecuencias(muestra_a, valores)
tabla_a
```

#Punto C

```python
resultados_c = []
for n in [100, 1000, 10000]:
    _, u = glc(n)
    muestra = transformada_inversa(u, [1,2], [1/3,2/3])
    resultados_c.append({
        "n": n,
        "Valores iguales a 1": int(np.sum(muestra == 1)),
        "Proporción observada": np.mean(muestra == 1),
        "Proporción teórica": 1/3
    })
pd.DataFrame(resultados_c)
```

#Punto D

```python
valores_d = np.arange(1,11)
prob_d = np.array([.06,.06,.06,.06,.06,.15,.13,.14,.15,.13])

p_grupo = np.array([prob_d[:5].sum(), prob_d[5:].sum()])
p_A = prob_d[:5] / p_grupo[0]
p_B = prob_d[5:] / p_grupo[1]

_, u_grupo = glc(100)
_, u_valor = glc(100)

muestra_d = np.empty(100, dtype=int)
A = u_grupo < p_grupo[0]
muestra_d[A] = transformada_inversa(u_valor[A], np.arange(1,6), p_A)
muestra_d[~A] = transformada_inversa(u_valor[~A], np.arange(6,11), p_B)

tabla_d = tabla_frecuencias(muestra_d, valores_d)
tabla_d["Probabilidad teórica"] = prob_d
tabla_d
```
